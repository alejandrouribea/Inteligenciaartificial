```python
# Carga de datos.
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('train.csv')
print(df.head())
```

```python
# Se muestra la cantidad de estudiantes por programa academico.
print("Cantidad de estudiantes por programa:")
print(df['ESTU_PRGM_ACADEMICO'].value_counts())


# Mostrar la gráfica
plt.show()
```

```python
# Visualización de estudiantes por periodo.
print("Cantidad de estudiantes por periodo:")
periodo_counts = df['PERIODO'].value_counts().sort_index()
print(periodo_counts)
periodo_counts = df['PERIODO'].value_counts().sort_index()

# Crear la gráfica
plt.figure(figsize=(10,6))
periodo_counts.plot(kind='bar', color='lightgreen')
plt.title('Cantidad de estudiantes por período')
plt.xlabel('Período')
plt.ylabel('Cantidad de estudiantes')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

# Mostrar la gráfica
plt.show()
```

```python
# Se muestra la cantidad de los cursos que se estudian por departamento.
print("Cantidad de los cursos que se estudian por departamento:")
df_cd = df.groupby(['ESTU_PRGM_DEPARTAMENTO', 'ESTU_PRGM_ACADEMICO']).count()
df_cd = df_cd['ID']
print (df_cd)
```

```python
# Cantidad de personas en los estractos 1,2,3,4,5,6
print("Cantidad de estudiantes en los diferentes estratos:")
estracto_counts = df['FAMI_ESTRATOVIVIENDA'].value_counts().sort_index()
print(estracto_counts)
estrato_counts = df['FAMI_ESTRATOVIVIENDA'].value_counts().sort_index()

# Crear la gráfica
plt.figure(figsize=(8,5))
estrato_counts.plot(kind='bar', color='salmon')
plt.title('Cantidad de estudiantes en los diferentes estratos')
plt.xlabel('Estrato')
plt.ylabel('Cantidad de estudiantes')
plt.xticks(rotation=0)
plt.tight_layout()

# Mostrar la gráfica
plt.show()
```

```python
# Cantidad de personas con y sin internet según el estrato.
print("Cantidad de estudiantes segun el estrato que cuentan con y sin internet:")
count_internet = df.groupby(['FAMI_ESTRATOVIVIENDA', 'FAMI_TIENEINTERNET']).size()
print(count_internet)
```

```python
# Porcentaje de personas que no pagan matrícula.
no_matricula = df[df['ESTU_VALORMATRICULAUNIVERSIDAD'] == 'No pagó matrícula'].shape[0]
total_estudiantes = df.shape[0]
porcentaje_no_matricula = (no_matricula / total_estudiantes) * 100
print("Porcentaje de personas que no pagan matrícula: ", porcentaje_no_matricula, "%")
```
